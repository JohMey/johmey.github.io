# Why electrons and holes can share a diffusion width
A light pulse makes equal excess electron and hole densities. A small internal electric field can make their particle fluxes agree even when their mobilities are very different. Compare this coupled motion with two independent Gaussian packets.

```python
import numpy as np
import matplotlib.pyplot as plt

T_K=300.
mu_n=.12             # m²/(V s)
mu_p=.04
n_over_p=1.0         # local equilibrium density ratio for the linearised coefficients
time_us=1.
initial_width_um=20.
assert min(T_K,mu_n,mu_p,n_over_p,initial_width_um)>0 and time_us>=0
```

## Impose vanishing total charge current
Take a slowly varying neutral excess with gradients equal for electrons and holes. Mobilities and background densities are uniform on the packet scale. The Einstein relation supplies diffusion constants for nondegenerate carriers.

```python
thermal_voltage=8.617333262e-5*T_K
Dn,Dp=mu_n*thermal_voltage,mu_p*thermal_voltage
Da=(mu_n*n_over_p*Dp+mu_p*Dn)/(mu_n*n_over_p+mu_p)
def width(D,t):
    return np.sqrt(initial_width_um**2+2*D*1e6*t)
print("Coupled diffusion coefficient (m²/s) and width (µm):", Da,width(Da,time_us))
assert min(Dn,Dp)<=Da+1e-15 and Da<=max(Dn,Dp)+1e-15
if n_over_p==1:assert np.isclose(Da,2*Dn*Dp/(Dn+Dp))
```

## Compare profiles and slope estimates
All plotted profiles have the same integrated excess. The spatial variance, rather than its square root, grows linearly. Include a longer time interval before estimating a slope from noisy data.

```python
fig,ax=plt.subplots(1,2,figsize=(9,3.9),layout="constrained")
xx=np.linspace(-4*width(max(Dn,Dp),time_us),4*width(max(Dn,Dp),time_us),300)
ts=np.linspace(0,max(2*time_us,.1),180)
for D,name,col in [(Dn,"Electrons alone","#375b9a"),(Dp,"Holes alone","#a23d21"),(Da,"Neutral packet","#175868")]:
    w=width(D,time_us)
    ax[0].plot(xx,np.exp(-.5*(xx/w)**2)/(np.sqrt(2*np.pi)*w),color=col,label=name)
    ax[1].plot(ts,width(D,ts)**2,color=col,label=name)
ax[0].set(xlabel="Position (µm)",ylabel="Normalised density (1/µm)",title="Profiles at the selected delay");ax[0].legend(fontsize=8)
ax[1].set(xlabel="Delay (µs)",ylabel="Spatial variance (µm²)",title="Extract the diffusion coefficient");ax[1].legend(fontsize=8)
plt.show()
```

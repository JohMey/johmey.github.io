# How can a field that decays in air transmit power to another prism?

Place a thin air gap between identical prisms. Derive the TE transmission from the two interface conditions and test when an exponential attenuation length predicts the actual throughput.

## Propagate both boundary data through the gap

A monochromatic TE wave in a prism of index $n$ has tangential wavenumber $k_x=nk_0\sin\theta$, normal wavenumber $k=nk_0\cos\theta$, and $k_0=2\pi/\lambda$. Above the critical angle the air equation becomes

$$
E''=\kappa^2 E,
$$

where $\kappa=(k_x^2-k_0^2)^{1/2}$. Matching $E$ and its derivative across each nonmagnetic interface retains both independent exponential solutions. Discarding the growing solution would preclude matching the outgoing wave in the second prism.

From the exit to the entrance the transfer matrix has diagonal entries $\cosh\kappa d$ and off-diagonal entries $-\sinh(\kappa d)/\kappa$ and $-\kappa\sinh(\kappa d)$. Insert $(t,ikt)$ at the exit and $(1+r,ik(1-r))$ at the entrance. Eliminating $r$ and taking the squared modulus gives the power transmission:

$$T=\left[1+\frac{(k^2+\kappa^2)^2}{4k^2\kappa^2}\sinh^2(\kappa d)\right]^{-1}.$$

Both exterior media are identical, so transmitted power fraction equals squared transmitted amplitude. The finite flux through the gap comes from interference between its two evanescent components. The single exponential at an isolated interface carries no average normal power.

## The penetration length sets a slope

At a large gap the logarithmic slope $-d\ln T/dd$ tends to $2\kappa$. At contact $T=1$ with zero slope, because the leading correction in gap thickness is quadratic. Measuring a straight-line slope over a very small gap would therefore return a biased estimate of the penetration depth. Use the first graph to choose the window for a fit and the second graph to assess the bias.
Take a vacuum wavelength of 600 nm, index 1.50 and incidence angle 50°. The critical angle is 41.81°. The amplitude penetration length $1/\kappa$ is 168.72 nm and the coefficient of $\sinh^2(\kappa d)$ is 1.312. For gaps of 50, 100 and 200 nm the transmitted fractions are 0.894, 0.659 and 0.257. Doubling the gap therefore does not simply square the measured transmission. Only the sufficiently thick-gap logarithmic slope approaches a constant.

One can invert a single transmission reading when the angle and indices are independently known. Write $A=(k^2+\kappa^2)^2/(4k^2\kappa^2)$. Then $d=\kappa^{-1}\operatorname{asinh}\sqrt{(T^{-1}-1)/A}$. The reading 0.659 returns a gap of about 100 nm. Near $T=1$ a small absolute power error makes a large relative error in the inferred gap. This follows directly from the quadratic expansion $1-T\simeq A(\kappa d)^2$.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
index = 1.5
angle_deg = 50.
wavelength_nm = 600.
```

```python
k0 = 2*np.pi/wavelength_nm
def coefficients(theta):
    angle=np.deg2rad(theta)
    return index*k0*np.cos(angle), k0*np.sqrt((index*np.sin(angle))**2-1)
def transmission(d,theta):
    k,decay=coefficients(theta)
    factor=(k*k+decay**2)**2/(4*k*k*decay**2)
    return 1/(1+factor*np.sinh(decay*d)**2)
critical=np.rad2deg(np.arcsin(1/index))
assert index > 1 and critical < angle_deg < 90 and wavelength_nm>0
k,decay=coefficients(angle_deg)
assert np.isclose(transmission(0,angle_deg),1)
print(f"Critical angle {critical:.2f}°, amplitude penetration depth {1/decay:.2f} nm")
```

```python
d=np.linspace(0,2*wavelength_nm,300)
fig,ax=axes("Air gap d (nm)","Transmitted power fraction")
for theta,colour in zip([critical+(90-critical)*.05,angle_deg,critical+(90-critical)*.40],[orange,teal,blue]):
    ax.semilogy(d,transmission(d,theta),label=f"{theta:.1f}°",color=colour)
ax.legend()
plt.show()
```

A higher transverse wavenumber causes a steeper thick-gap decay. At contact every angle has unit transmission.

```python
slope=-np.gradient(np.log(transmission(d,angle_deg)),d,edge_order=2)
fig,ax=axes("Dimensionless gap κd","Local attenuation slope / (2κ)")
ax.plot(decay*d,slope/(2*decay),color=teal,label="Finite gap")
ax.axhline(1,color=orange,ls="--",label="Thick-gap limit")
ax.legend()
plt.show()
```

Normalising by twice the decay constant turns the thick-gap limit into one. The early points reveal why a thin-gap slope fit fails.

## Exercise · halve a thick-gap signal

How much extra separation is required to halve the transmitted power deep in the exponential regime?

The ratio tends to $\exp(-2\kappa\Delta d)$. Setting it to one half gives

$$
\Delta d=\ln 2/(2\kappa).
$$

This uses a power attenuation length. Confusing amplitude and intensity introduces a factor of two.

## Exercise · choose a boundary at the far side

What changes if the second prism is removed while the incident prism and angle remain the same?

The gap becomes a half-space and boundedness removes the growing component. The transmitted evanescent amplitude can still be finite at the interface, while the average normal power flux becomes zero. The limit of the two-prism transmission agrees as $d$ tends to infinity.

# Why does a Helmholtz pair have a useful uniform region?

A magnetometer averages the field over a finite sample. Derive the coil spacing that suppresses curvature and predict how rapidly the remaining calibration error grows away from the midpoint.

## Cancel a spatial derivative with geometry

For an ideal circular loop, Biot–Savart gives the axial field $B(z)=\mu_0NI R^2/[2(R^2+z^2)^{3/2}]$. It follows by adding equal axial components of the current elements around the circumference. Place two identical loops at $\pm s/2$, carrying currents in the same sense. Superposition and mirror symmetry remove every odd derivative at the midpoint.

Differentiate the single-loop shape $f(u)=(R^2+u^2)^{-3/2}$ twice. The result is $3(4u^2-R^2)(R^2+u^2)^{-7/2}$. The second derivative of the pair changes sign at $s=R$. At this spacing both the linear and quadratic spatial errors vanish. Expanding the even sum one order farther gives

$$\frac{B(z)}{B(0)}=1-\frac{144}{125}\left(\frac zR\right)^4+O((z/R)^6).$$

Flatness concerns a region of space. A high field value at the centre alone cannot establish that a sample extending on both sides was excited uniformly.

## Set the acceptance criterion before choosing a coil

If the allowed pointwise fractional variation is $10^{-3}$, the asymptotic half-length of the uniform region is $R(125\times10^{-3}/144)^{1/4}\simeq0.172R$. With a 100 mm radius this suggests 17 mm on either side. Compare that estimate with the exact field in the notebook. Small spacing errors restore the quadratic term. The accuracy of the coil fixture can therefore matter more than the current-meter resolution when calibrating an extended sample.
At 100 ampere-turns per coil and $R=0.100$ m, the central field is $(4/5)^{3/2}\mu_0 NI/R=0.899$ mT. A point 10 mm off centre experiences a leading relative deviation of $1.152\times10^{-4}$, about 0.104 µT in absolute field. A Hall probe with resolution 1 µT would struggle to distinguish a single such deviation. The pointwise specification and the spatial resolution of a verification scan therefore belong in the same calculation.

For a uniformly sensitive sample extending from $-L$ to $L$, average the quartic term instead of substituting the outermost coordinate. Since the average of $z^4$ is $L^4/5$, the leading bias of its mean response is a fifth of the edge deviation. A sample that fills the nominal tolerance region can report an apparently excellent mean field while a process sensitive to the edge already operates at the specified limit.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
radius_m = 0.1
spacing_ratio = 1.0
ampere_turns_per_coil = 100.0
```

```python
mu0 = 4e-7*np.pi
def bfield(z, ratio):
    d = ratio*radius_m/2
    return mu0*ampere_turns_per_coil*radius_m**2/2 * (
        (radius_m**2+(z-d)**2)**-1.5 + (radius_m**2+(z+d)**2)**-1.5)
assert radius_m>0 and ampere_turns_per_coil>0 and spacing_ratio>0
assert np.isclose(bfield(0,1),mu0*ampere_turns_per_coil/radius_m*(4/5)**1.5)
print(f"Centre field: {bfield(0,spacing_ratio)*1e3:.4f} mT")
```

```python
u = np.linspace(-.7,.7,301)
fig, ax = axes("Axial position z/R", "Field divided by central field")
for ratio,colour in zip([.9,1.,1.1],[orange,teal,blue]):
    ax.plot(u,bfield(u*radius_m,ratio)/bfield(0,ratio),label=f"s/R = {ratio:g}",color=colour)
ax.legend()
plt.show()
```

Offset the separation by a tenth of the radius in each direction. The signs of the resulting calibration errors reverse close to the centre.

```python
u = np.geomspace(.006,.5,140)
fig, ax = axes("Distance from centre |z|/R", "Absolute relative field deviation")
ax.loglog(u,abs(bfield(u*radius_m,spacing_ratio)/bfield(0,spacing_ratio)-1),label="Exact coil pair",color=teal)
ax.loglog(u,144*u**4/125,"--",label="Ideal quartic term",color=orange)
ax.legend()
plt.show()
```

At ideal spacing the error falls sixteenfold when the sample half-length halves. Move the spacing off its design value to recover quadratic scaling.

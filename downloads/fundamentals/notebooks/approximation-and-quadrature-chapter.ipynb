# Three Gaussian-weighted samples of a spectral response
A dimensionless variable has probability density exp(−x²)/√π. Estimate its mean response to cos(ωx) by a three-node Hermite rule, then compare with the exact Gaussian Fourier transform. Vary the carrier and see whether exact low moments protect every observable.

```python
import numpy as np
import matplotlib.pyplot as plt
omega = 2.0
assert 0 <= omega <= 10
nodes=np.array([-np.sqrt(1.5),0,np.sqrt(1.5)])
weights=np.array([1,4,1])/6
def estimate(w):
    return np.sum(weights*np.cos(np.asarray(w)[...,None]*nodes),axis=-1)
for degree,value in enumerate([1.,0.,.5,0.,.75,0.]):
    assert np.isclose(weights@(nodes**degree),value)
exact=lambda w:np.exp(-np.asarray(w)**2/4)
print(f"Three samples {estimate(omega):.8f}, exact response {exact(omega):.8f}")
```

## Inspect what the nodes measure
The curve on the left gives the density-weighted integrand. The vertical lines locate the Hermite sampling coordinates. The right panel shows the exact and discrete responses over a wider range of frequency.

```python
fig,(shape,response)=plt.subplots(1,2,figsize=(10,4),layout="constrained")
x=np.linspace(-3.2,3.2,450)
shape.plot(x,np.exp(-x*x)*np.cos(omega*x)/np.sqrt(np.pi),color="#145c71")
for n in nodes:shape.axvline(n,color="#b94626",ls=":",lw=1.5)
shape.set(xlabel="x in Gaussian scale units",ylabel="Weighted integrand",title=f"Three sampling locations at ω = {omega:g}")
ww=np.linspace(0,10,401)
response.plot(ww,exact(ww),color="#145c71",label="continuous Gaussian")
response.plot(ww,estimate(ww),color="#b94626",ls="--",label="three nodes")
response.plot(omega,estimate(omega),"o",color="#b94626")
response.set(xlabel="Carrier frequency ω",ylabel="Mean cosine response",title="Moment exactness has a frequency range")
response.legend()
for ax in (shape,response):ax.grid(alpha=.2)
plt.show()
```

## Independent reference
Use NumPy's higher-order Gauss–Hermite rule and remove its unnormalised weight √π. At large frequencies three nodes cannot resolve phase cancellations.

```python
xh,wh=np.polynomial.hermite.hermgauss(64)
dense=np.sum(wh*np.cos(omega*xh))/np.sqrt(np.pi)
assert abs(dense-exact(omega))<2e-12
print("64-node check:",dense)
```

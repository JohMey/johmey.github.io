# Locate the load on a submerged viewing window

Two circular windows have the same area and centre depth. One is more steeply inclined. Predict whether the net force and its position both stay the same. The circular face is completely submerged and the opposite face is at atmospheric pressure.

```python
import numpy as np
import matplotlib.pyplot as plt

radius,depth=.18,.42   # m, circular radius and vertical centre depth
tilt=60.               # degrees from horizontal
rho,g=998.,9.81        # kg/m³, m/s²
assert radius>0 and rho>0 and g>0 and 0<=tilt<=90
sine=np.sin(np.deg2rad(tilt))
assert depth>radius*sine, "This worksheet assumes a fully submerged window."
```

## Calculate both the resultant and the moment
The first moment of area about the centre is zero. The second moment about an in-plane diameter is pi times the fourth power of radius divided by four. Integrate horizontal chord strips to check the resulting force location independently.

```python
A=np.pi*radius**2
force=rho*g*A*depth
offset=radius**2*sine/(4*depth)
edges=np.linspace(-radius,radius,3001)
s=(edges[1:]+edges[:-1])/2
area=2*np.sqrt(radius**2-s*s)*np.diff(edges)
pressure=rho*g*(depth+s*sine)
numerical=(pressure*area).sum()
s_numeric=(pressure*s*area).sum()/numerical
assert np.isclose(force,numerical,rtol=3e-5) and np.isclose(offset,s_numeric,atol=1e-5*radius)
print(f"Resultant {force:.2f} N, offset {1000*offset:.2f} mm down the slope.")
```

## Inspect which strips produce the moment
Pressure rises linearly with depth but strip width vanishes at the rim. The right panel holds depth fixed as the angle varies and only draws submerged orientations.

```python
fig,ax=plt.subplots(1,2,figsize=(9,3.5),layout="constrained")
ax[0].plot(s*1000,pressure*2*np.sqrt(radius**2-s*s),label="Force per unit slope distance")
ax[0].axvline(offset*1000,color="#ab532c",ls="--",label="Resultant location")
ax[0].set(xlabel="Distance down window from centre (mm)",ylabel="Strip force density (N/m)",title="Integrated hydrostatic load")
theta=np.linspace(0,90,120); valid=depth>radius*np.sin(np.deg2rad(theta))
ax[1].plot(theta[valid],1000*radius**2*np.sin(np.deg2rad(theta[valid]))/(4*depth),label="Centre of pressure")
ax[1].set(xlabel="Tilt from horizontal (degrees)",ylabel="Down-slope offset (mm)",title="Location at fixed centre depth")
for p in ax: p.grid(alpha=.22);p.legend(fontsize=8)
plt.show()
```

# Which electrode sees a charge moving above a detector?

A point charge moves without touching the electrodes. Derive the induced charge on one strip from a second, source-free boundary problem. Use the two-dimensional map to distinguish motion across the detector from motion away from it.

## Exchange the source and the measuring electrode

An infinite conducting plane is split into long parallel strips. The gaps have negligible width and every strip is held at zero potential during the measurement. Put one charge $q$ at $(x,z)$ above the plane. Computing the field separately for every position would be expensive. Instead consider an auxiliary problem with the measuring strip at unit potential, all other strips grounded and no charge above the plane. Call its dimensionless solution $w$.

Green's reciprocity pairs the electrode potentials of one problem with the charges of the other. The grounded real electrodes make one boundary contribution vanish. The test electrode gives $Q_{\rm strip}$ and the real point source gives $q w(x,z)$. Their sum vanishes, hence $Q_{\rm strip}=-qw$. For a strip occupying $-a<u<a$, the Poisson kernel of the upper half-plane integrates the boundary value into the interior:

$$w(x,z)=\frac{1}{\pi}\int_{-a}^{a}\frac{z\,du}{(u-x)^2+z^2}
=\frac{\arctan[(a-x)/z]+\arctan[(a+x)/z]}{\pi}.$$

This weighting potential is computed for the electrode geometry. It is independent of the test charge. Its gradient predicts the readout current as a charge travels through different weightings. Infinite strip length and an effectively infinite return plane are the assumptions that permit this two-dimensional construction.

## Read a charge and a displacement separately

At the centre and at height $z=a$, the strip subtends a right angle, giving $w=1/2$. A positive charge of 8 fC induces −4 fC on this electrode. At $x=a,z=a$ the weight is $\arctan(2)/\pi$, about 0.352. The change in measured charge can occur with the same charge and the same height. The peak and width of a lateral scan together carry more positional information than one sample. A finite insulating gap or a grounded back plane would change the boundary problem.
A second measurement illustrates why electrode geometry matters for a current signal. Differentiate the central response $w(0,z)=2\arctan(a/z)/\pi$, giving $\partial_z w=-2a/[\pi(a^2+z^2)]$. At $z=a=1$ mm, an 8 fC charge moving upward at 1 mm/µs changes the induced charge at $+8/\pi$ fC/µs, or $+2.55$ nA. The sign here describes the time derivative of charge on the strip. The sign of an external current reading also depends on the chosen lead direction.

Summing the strip weights over an entire tiled plane gives one because the Poisson kernel integrates to one. Total induced charge on that plane is therefore $-q$, even as charge redistributes between individual electrodes. A finite numerical map can miss the long tails of the weighting function. Enlarge the lateral range and check the sum before interpreting a shortfall as an uncollected signal.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
half_width_mm = 1.0
charge_fc = 8.0
height_mm = 1.0
```

```python
def weighting(x, z):
    a = half_width_mm
    return (np.arctan((a - x)/z) + np.arctan((a + x)/z))/np.pi

assert half_width_mm > 0 and height_mm > 0
assert np.isclose(weighting(0, half_width_mm), 0.5)
print(f"Induced charge on the central strip: {-charge_fc*weighting(0,height_mm):.4f} fC")
```

```python
a = half_width_mm
x = np.linspace(-3*a, 3*a, 241)
z = np.linspace(0.02*a, 3*a, 121)
X, Z = np.meshgrid(x, z)
fig, ax = axes("Lateral coordinate x (mm)", "Height z (mm)")
cs = ax.contourf(X, Z, weighting(X,Z), levels=np.linspace(0,1,21), cmap="viridis")
fig.colorbar(cs, ax=ax, label="Weighting potential")
ax.plot([-a,a],[0,0], lw=4, color=orange)
plt.show()
```

Contours show equal sensitivity to a point charge. The steepest change occurs close to an electrode edge.

```python
fig, ax = axes("Lateral coordinate x (mm)", "Induced electrode charge (fC)")
for multiplier, colour in zip([0.5,1,2], [orange,teal,blue]):
    h = height_mm*multiplier
    ax.plot(x, -charge_fc*weighting(x,h), label=f"z = {h:g} mm", color=colour)
ax.legend()
plt.show()
```

Move the charge at constant height and record the induced charge. The scan width can help distinguish a small nearby charge from a larger distant one.

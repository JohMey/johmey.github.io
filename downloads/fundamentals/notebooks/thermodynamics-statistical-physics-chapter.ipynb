# A leaky vessel changes its mixture
Two nonreacting species share a vessel at fixed temperature. Their pressures initially add according to the ideal-gas law, but the molecules hitting a small hole have a speed bias. Predict whether the gas left behind or the gas escaping initially is richer in the light species.

```python
import numpy as np
import matplotlib.pyplot as plt

light_mass_u=20.
heavy_mass_u=40.
light_initial=.50
heavy_time_s=100.      # heavy-species depletion time V/(A <vz positive>)
assert 0<light_mass_u<heavy_mass_u and 0<light_initial<1 and heavy_time_s>0
```

## Integrate the separate number balances
The ratio of depletion rates follows the inverse square root of mass. Use two time constants to compute the composition inside the vessel, in the instantaneous beam, and in the collection integrated since the start.

```python
rate_ratio=np.sqrt(heavy_mass_u/light_mass_u)
t=np.linspace(0,3*heavy_time_s,241)
a=light_initial*np.exp(-rate_ratio*t/heavy_time_s)
b=(1-light_initial)*np.exp(-t/heavy_time_s)
remaining=a/(a+b); beam=rate_ratio*a/(rate_ratio*a+b)
escaped_a=light_initial*(-np.expm1(-rate_ratio*t/heavy_time_s))
escaped_b=(1-light_initial)*(-np.expm1(-t/heavy_time_s))
collected=np.divide(escaped_a,escaped_a+escaped_b,out=np.full_like(a,beam[0]),where=t>0)
print("Initial light fraction in vessel and escape flux:", remaining[0],beam[0])
assert np.allclose(a+escaped_a,light_initial) and np.allclose(b+escaped_b,1-light_initial)
assert np.all(beam>=remaining)
```

## Compare count rates with collected composition
At very late times the instantaneous beam becomes mostly heavy even though the first collected atoms were enriched in the light species. Collecting nearly every original atom must eventually reproduce the initial composition.

```python
fig,ax=plt.subplots(1,2,figsize=(9,3.9),layout="constrained")
ax[0].plot(t,a,color="#375b9a",label="Light species");ax[0].plot(t,b,color="#a23d21",label="Heavy species")
ax[0].set(xlabel="Time (s)",ylabel="Number / initial total",title="Separate reservoir balances");ax[0].legend(fontsize=8)
for y,label,c in [(remaining,"In vessel","#375b9a"),(beam,"Instantaneous flux","#a23d21"),(collected,"Collected since start","#175868")]:ax[1].plot(t,y,label=label,color=c)
ax[1].set(xlabel="Time (s)",ylabel="Light-species fraction",title="Three different measurements");ax[1].legend(fontsize=8)
plt.show()
```

# Can two ohmic layers imitate a relaxing dielectric?

Model a film and its contact layer in series. Derive the frequency scale of interfacial charge accumulation and decide which features of a measured spectrum can arise from geometry alone.

## Match total current at a leaky interface

Take harmonic fields proportional to $e^{i\omega t}$. In layer $j$, Ohm's law and Maxwell's displacement current give $J=Y_j E_j$ with $Y_j=\sigma_j+i\omega\epsilon_j$. Use absolute permittivity in SI units. Total current is continuous across the interface. If the first layer occupies thickness fraction $f$, the imposed mean field is

$$
\bar E=fE_1+(1-f)E_2.
$$

Eliminating the layer fields yields

$$Y_{\rm eff}=\frac{Y_1Y_2}{fY_2+(1-f)Y_1},\qquad
\tau=\frac{f\epsilon_2+(1-f)\epsilon_1}{f\sigma_2+(1-f)\sigma_1}.$$

The denominator is proportional to $1+i\omega\tau$. A pole has appeared even though each material has frequency-independent conductivity and permittivity. The free interfacial charge per unit area is $\epsilon_2 E_2-\epsilon_1 E_1$ with the normal from layer 1 to layer 2. At low frequency the less conducting layer takes a larger field. At high frequency the displacement currents determine the field sharing.

## Separate dielectric storage from direct conduction

Infer the relative storage permittivity as $\operatorname{Im}(Y_{\rm eff})/(\omega\epsilon_0)$. A DC current is real and drops out of this quantity. For layers with the same relative permittivity, their high-frequency effective value must equal that common value. In the DC limit the effective conductivity is the thickness-weighted harmonic mean. These two checks are independent. The interface can nevertheless produce a large rise in apparent storage on the low-frequency side.
Use a first layer that fills 20% of the gap and has conductivity $10^{-8}$ S/m. Give the other layer conductivity $10^{-6}$ S/m and both layers relative permittivity 4. The DC conductivity is $4.81\times10^{-8}$ S/m. An applied mean field of 1000 V/m therefore drives $4.81\times10^{-5}$ A/m². The weakly conducting layer needs about 4808 V/m to carry it and the second layer only 48.1 V/m. Their thickness-weighted mean recovers the imposed field.

The time constant is 0.170 ms, corresponding to 935 Hz. Expanding the two complex conductivities to first order in frequency gives the low-frequency relative storage permittivity $4[f\sigma_2^2+(1-f)\sigma_1^2]/[f\sigma_2+(1-f)\sigma_1]^2=18.50$. At high frequency it falls to 4. Make the conductivities equal in the notebook and predict the interfacial charge before running it. The denominator still defines a time scale algebraically but its pole cancels out of the measured admittance.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
fraction = .20
sigma1, sigma2 = 1.e-8, 1.e-6   # S/m
er1, er2 = 4., 4.
mean_field = 1000.                # V/m
```

```python
eps0=8.8541878128e-12
eps1,eps2=er1*eps0,er2*eps0
assert 0<fraction<1 and min(sigma1,sigma2,eps1,eps2)>0
tau=(fraction*eps2+(1-fraction)*eps1)/(fraction*sigma2+(1-fraction)*sigma1)
f_relax=1/(2*np.pi*tau)
def layers(freq):
    w=2*np.pi*np.asarray(freq)
    y1,y2=sigma1+1j*w*eps1,sigma2+1j*w*eps2
    eff=1/(fraction/y1+(1-fraction)/y2)
    return eff,eff*mean_field/y1,eff*mean_field/y2
assert np.isclose(layers(0)[0].real,1/(fraction/sigma1+(1-fraction)/sigma2),atol=1e-16)
print(f"Interfacial relaxation frequency {f_relax:.2f} Hz")
```

```python
freq=f_relax*np.geomspace(.001,1000,300)
eff,e1,e2=layers(freq)
fig,ax=axes("Frequency (Hz)","Apparent storage permittivity / ε₀")
ax.semilogx(freq,eff.imag/(2*np.pi*freq*eps0),color=teal)
ax.axvline(f_relax,color=orange,ls="--",label="Interfacial relaxation")
ax.legend()
plt.show()
```

Use the measured transition frequency and a thickness series to distinguish interfacial storage from an intrinsic molecular relaxation.

```python
fig,ax=axes("Frequency (Hz)","Layer field amplitude / imposed amplitude")
ax.semilogx(freq,abs(e1/mean_field),label="Layer 1",color=teal)
ax.semilogx(freq,abs(e2/mean_field),label="Layer 2",color=orange)
ax.legend()
assert np.allclose(fraction*e1+(1-fraction)*e2,mean_field)
plt.show()
```

Each curve is an amplitude. Add the complex fields with their phases before checking that the thickness-weighted average equals the applied field.

## Exercise · remove the relaxation

Which choice of layer properties makes the interfacial charge vanish for every frequency?

The charge vanishes if $\epsilon_2/Y_2=\epsilon_1/Y_1$. The imaginary terms cancel identically after cross multiplication, leaving $\epsilon_2\sigma_1=\epsilon_1\sigma_2$. The two dielectric relaxation times are then equal. Check the limit by making both layers identical.

## Exercise · exchange the materials

Relabel the layers and their thickness fractions. Which measurements can distinguish the order of this stack?

The two-terminal admittance is unchanged under $(1,2,f)\mapsto(2,1,1-f)$. It cannot identify which material is nearer the upper electrode. A local field measurement at a known depth or a contact-specific measurement can distinguish the configurations.

# How does a converging oil film support a load?

Two plates have a small fixed angle. The lower plate moves while the upper pad stands still. Derive the pressure from volume conservation and local shear, then change film height and speed to see why a wedge develops lift.

## Combine pressure flow and wall-driven flow before imposing continuity

Let $x$ point along the moving surface from the wide inlet $h_1$ to the thin outlet $h_2$. The film has constant density and viscosity $\mu$. It is shallow compared with its length $L$, and the pad is broad enough to neglect side leakage. Local steady momentum balance is $\mu u_{yy}=dp/dx$. Integrate across the gap with $u(0)=U$ and $u(h)=0$. The linear boundary-driven velocity combines with a parabolic pressure-driven term. Integration of that velocity gives the volume flow per unit width
$$q=Uh/2-\frac{h^3}{12\mu}\frac{dp}{dx}.$$
The same $q$ passes every station. Introduce $h_*=2q/U$, making $dp/dx=6\mu U(h-h_*)/h^3$. Equal ambient gauge pressures at both ends require the integral of the gradient over $x$ to vanish. For a linearly tapering film convert integration from $x$ to $h$ and obtain $h_*=2h_1h_2/(h_1+h_2)$. The pressure maximum lies where $h=h_*$. Integrating once more gives a direct pressure profile:
$$p=\frac{6\mu UL}{h_1-h_2}\left[\frac1h-\frac1{h_1}-\frac{h_*}{2}\left(\frac1{h^2}-\frac1{h_1^2}\right)\right].$$
This form subtracts nearby terms in nearly parallel gaps. The notebook uses the equivalent factored result $p=6\mu U L(h_1-h_2)s(1-s)/[(h_1+h_2)h^2]$, where $s=x/L$. It remains well behaved as the slope tends to zero.

## Measure the pressure gradient where the Couette flux matches the actual flow

Choose inlet and outlet heights of 120 and 60 µm, length 40 mm, sliding speed 1 m/s and viscosity 0.050 Pa s. Then $h_*=80$ µm and the flux per unit width is $4.00\times10^{-5}$ m²/s. The pressure maximum is two thirds of the length from the inlet. For these data it is approximately 0.139 MPa. At that point the pressure-driven part of the flow vanishes locally even though the accumulated pressure above ambient is greatest. This is the distinction between a local gradient and an absolute load-carrying pressure.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
hin_um, hout_um = 120., 60.
L_mm, speed, viscosity = 40., 1., .05
```

```python
hin,hout,L=hin_um*1e-6,hout_um*1e-6,L_mm/1000
assert hin>=hout>0 and min(L,speed,viscosity)>0
def pressure(s, hi, ho):
    h=hi+(ho-hi)*s
    return 6*viscosity*speed*L*(hi-ho)*s*(1-s)/((hi+ho)*h*h)
s=np.linspace(0,1,501); h=hin+(hout-hin)*s
hstar=2*hin*hout/(hin+hout)
q=speed*hstar/2
p=pressure(s,hin,hout)
pos=hin/(hin+hout)
load=L*np.sum(np.diff(s)*(p[1:]+p[:-1])/2)
assert np.allclose(p[[0,-1]],0) and np.all(p>=0)
print(f"Peak {pressure(pos,hin,hout)/1e6:.4f} MPa at x/L = {pos:.4f}; load per width {load:.1f} N/m.")
```

```python
fig,ax=axes("Position x/L","Gauge pressure (MPa)")
for ratio,col in [(1,blue),(hin/hout,teal),(3,orange)]:
    ax.plot(s,pressure(s,ratio*hout,hout)/1e6,color=col,label=f"Inlet / outlet = {ratio:.2f}")
ax.legend();plt.show()
```

The outlet thickness is held fixed in this comparison. A parallel gap carries fluid by shear while producing zero pressure load under the stated boundary conditions.

```python
fig,ax=axes("Position x/L","Volume flux per width (mm²/s)")
wall=speed*h/2
ax.plot(s,1e6*wall,label="Wall-driven contribution",color=orange)
ax.plot(s,1e6*(q-wall),label="Pressure-driven contribution",color=blue)
ax.axhline(q*1e6,color=teal,ls="--",label="Total flux")
ax.legend();plt.show()
```

The pressure-driven contribution opposes motion near the wide inlet and assists it nearer the outlet. Its sign changes at the pressure maximum.

## Exercise · scale both clearances

Double inlet and outlet thickness while holding taper ratio, viscosity, length and speed fixed. What becomes of the pressure and the flow?

The factor $(h_1-h_2)/(h_1+h_2)$ is constant, while the squared local height quadruples. Pressure and load per width fall to one quarter. The harmonic-mean height $h_*$ doubles, so the through-flow doubles.

## Exercise · interpret a reversed pad

What does the algebra predict for a diverging gap with the same wall direction? Under what condition does that extrapolation fail?

The sign of $h_1-h_2$ reverses, giving negative gauge pressure between the equal-pressure boundaries. Small subambient pressures can exist in a filled liquid. Once the absolute pressure reaches the cavitation threshold or gas enters the film, a full-film solution cannot represent the cavity or its mass balance. The notebook therefore keeps its default sweep in the converging regime.

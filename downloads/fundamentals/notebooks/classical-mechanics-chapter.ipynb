# When an unresolved binary looks stationary
Change the companion's mass and brightness independently. The detector sees the centre of light in a projected circular orbit; the centre of mass remains fixed. Decide in advance when the astrometric signal will vanish.

```python
import numpy as np
import matplotlib.pyplot as plt

mass_fraction = .30       # M2 / (M1 + M2)
light_fraction = .10      # L2 / (L1 + L2)
separation_au = 1.0
distance_pc = 20.
inclination_deg = 55.
assert 0<mass_fraction<1 and 0<=light_fraction<=1 and distance_pc>0 and separation_au>0
```

## Construct the sky projection
One astronomical unit at one parsec subtends one arcsecond. Use milliarcseconds (mas). Inclination foreshortens the minor axis but preserves the major semiaxis of a projected circular orbit.

```python
phase=np.linspace(0,2*np.pi,181); theta=1000*separation_au/distance_pc
relative=theta*np.array([np.cos(phase),np.cos(np.deg2rad(inclination_deg))*np.sin(phase)])
p1=-mass_fraction*relative; p2=(1-mass_fraction)*relative
centroid=(1-light_fraction)*p1+light_fraction*p2
print("Centre-of-light semimajor axis (mas):",theta*abs(light_fraction-mass_fraction))
assert np.allclose((1-mass_fraction)*p1+mass_fraction*p2,0)
assert np.allclose(centroid,(light_fraction-mass_fraction)*relative)
```

## Observe the sign flip as the brightness ratio changes
A detector resolving only the photocentre cannot recover component separation without information about the component fractions. The second panel keeps an orientation sign that an amplitude alone would discard.

```python
fig,ax=plt.subplots(1,2,figsize=(9,3.9),layout="constrained")
for position,name,colour in [(p1,"Primary","#375b9a"),(p2,"Companion","#a23d21"),(centroid,"Centre of light","#175868")]:
    ax[0].plot(*position,label=name,color=colour);ax[0].plot(*position[:,0],"o",ms=4,color=colour)
ax[0].set(xlabel="Sky x (mas)",ylabel="Sky y (mas)",aspect="equal",title="Projected circular orbits");ax[0].legend(fontsize=8)
light=np.linspace(0,1,120);ax[1].plot(light,theta*(light-mass_fraction),color="#175868");ax[1].axhline(0,color="grey",lw=.8)
ax[1].plot(light_fraction,theta*(light_fraction-mass_fraction),"o",color="#a23d21")
ax[1].set(xlabel="Companion light fraction",ylabel="Signed centroid semiaxis (mas)",title="Bias caused by unresolved light")
plt.show()
```

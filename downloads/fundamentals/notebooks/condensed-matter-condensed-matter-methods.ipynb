# Read twist and mismatch from the moiré of two lattices

Stack two square grids that differ slightly in orientation or pitch. The slowly varying registry map is visible over many unit cells. Derive its scale and direction in reciprocal space, and find which combinations of layer parameters a single period measurement can resolve.

## Subtract two spatial phases

One family of planes in layer one carries a density modulation proportional to $\cos(\mathbf G_1\cdot\mathbf r)$, with $|\mathbf G_1|=2\pi/a$. Layer two has spacing $a(1+\delta)$ and is rotated by $\theta$. Multiplying their modulations, as in a local registry correlation, produces half the sum of two cosines. The rapidly varying term contains $\mathbf G_1+\mathbf G_2$; averaging over several original cells leaves the term with $\Delta\mathbf G=\mathbf G_1-\mathbf G_2$. A distance parallel to $\Delta\mathbf G$ that advances the resulting phase by $2\pi$ spans one full registry period.

Taking the dot product of the rotated reciprocal vectors gives the exact geometric scale

$$\frac{\Lambda}{a}=\left[1+\frac{1}{(1+\delta)^2}-\frac{2\cos\theta}{1+\delta}\right]^{-1/2}. $$

The second perpendicular family of a square lattice rotates and rescales the same way. Its difference vector is perpendicular to the first and has the same magnitude. Their two slow cosines generate a square registry pattern. For small mismatch and small twist measured in radians the period is approximately $a/\sqrt{\delta^2+\theta^2}$. The interference construction specifies a local continuum period. An exact common atomistic supercell would require additional commensurability conditions.

## Interpret a long measured period

Take $a=0.246$ nm, mismatch 1.3 percent and twist two degrees. In units of $2\pi/a$ the horizontal family's difference vector is $(1-\cos\theta/(1+\delta),-\sin\theta/(1+\delta))$, approximately $(0.0134,-0.0345)$. Its length is about 0.0370, so the registry period reaches about 6.65 nm. Each layer still has a spacing near a quarter nanometre. The larger scale is the distance over which small phase differences accumulate.

If an image measures only this period and the first layer spacing, it supplies a single scalar for the two unknowns $\delta$ and $\theta$. A reciprocal-space measurement referenced to the first lattice supplies direction as well. In the simple isotropic model this resolves the rotated and rescaled vector by $\mathbf G_2=\mathbf G_1-\Delta\mathbf G$. Additional distortions or strain require more directions before an inversion is possible.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
spacing_nm = .246
mismatch = .013
twist_deg = 2.
```

```python
assert spacing_nm>0 and mismatch>-1
angle = np.deg2rad(twist_deg)
g = 2*np.pi/spacing_nm
g1 = np.array([g,0.])
g2 = g*np.array([np.cos(angle),np.sin(angle)])/(1+mismatch)
difference = g1-g2
norm = np.linalg.norm(difference)
period = 2*np.pi/norm if norm>1e-12*g else np.inf
orthogonal = np.array([-difference[1],difference[0]])
assert np.isclose(np.dot(difference,orthogonal),0,atol=1e-10)
print(f"Registry period {period:.5g} nm; difference vector {difference} per nm")
```

```python
fig,(ax,zoom) = plt.subplots(1,2,figsize=(7,4),constrained_layout=True)
def arrow(axis,start,vector,colour):
    axis.annotate("",xy=start+vector,xytext=start,arrowprops=dict(arrowstyle="->",color=colour,lw=2))
for vec,colour,label in [(g1,teal,"Layer 1"),(g2,orange,"Layer 2")]:
    arrow(ax,np.zeros(2),vec,colour)
    second = np.array([-vec[1],vec[0]])
    arrow(ax,np.zeros(2),second,colour)
    ax.plot(vec[0],vec[1],"o",color=colour,label=label)
    ax.plot(second[0],second[1],"o",color=colour)
full_vectors=np.array([g1,g2,[0,g],[-g2[1],g2[0]],[0,0]])
low=full_vectors.min();high=full_vectors.max();pad_full=.12*(high-low)
ax.set(xlim=(low-pad_full,high+pad_full),ylim=(low-pad_full,high+pad_full))
arrow(zoom,g2,difference,blue);zoom.scatter([g1[0],g2[0]],[g1[1],g2[1]],c=[teal,orange])
pad=max(norm*.7,g*.02);centre=(g1+g2)/2
zoom.set(xlim=(centre[0]-pad,centre[0]+pad),ylim=(centre[1]-pad,centre[1]+pad))
for axis,heading in [(ax,"Reciprocal vectors"),(zoom,"Difference at the tips")]:
    axis.set(xlabel="Gx (nm⁻¹)",ylabel="Gy (nm⁻¹)",title=heading);axis.set_aspect("equal",adjustable="box");axis.grid(alpha=.2)
ax.legend(loc="best",fontsize=8)
plt.show()
```

The right panel enlarges the displacement between reciprocal-vector tips. Its direction is the normal to one family of registry fringes.

```python
extent = 1.5*period if np.isfinite(period) else 10*spacing_nm
x = np.linspace(-extent,extent,251)
X,Y = np.meshgrid(x,x)
registry = (np.cos(difference[0]*X+difference[1]*Y)+np.cos(orthogonal[0]*X+orthogonal[1]*Y))/2
fig,ax=axes("Horizontal distance (nm)","Vertical distance (nm)")
image=ax.imshow(registry,origin="lower",extent=(-extent,extent,-extent,extent),vmin=-1,vmax=1,cmap="Greys")
ax.grid(False);fig.colorbar(image,ax=ax,label="Registry correlation")
plt.show()
```

The map sums the two perpendicular slow modulations; the colour bar measures their dimensionless average. At perfect alignment with equal spacing the registry is constant everywhere.

## Exercise · determine a twist from pure rotation

The lattice spacings have been verified equal at 0.246 nm. What is the magnitude of the twist if the registry spacing is 10.0 nm?

With $\delta=0$, the reciprocal vectors form an isosceles triangle and $\Lambda=a/[2|\sin(\theta/2)|]$. The small positive solution is $2\arcsin(0.246/20.0)=0.02460$ rad or 1.41 degrees. Changing the twist sign reflects the pattern without changing its period.

## Exercise · make the same spacing with no twist

Use the same first-layer spacing and measured 10.0 nm period. What positive mismatch would explain the observation for aligned layers?

At $\theta=0$ and positive mismatch, $\Lambda=a(1+\delta)/\delta$. Solving gives $\delta=a/(\Lambda-a)=0.02522$, about 2.52 percent. Comparing this result with the previous exercise shows explicitly why period alone cannot identify a twist.

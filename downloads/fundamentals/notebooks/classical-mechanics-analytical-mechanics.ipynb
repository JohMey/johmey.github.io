# Which path is shortest across the surface of a cone?

A point can move along a conical surface while its azimuth and height both change. Unroll the geometry and predict the shortest path between two marked points. The measurable minimum radius distinguishes the geodesic from a line drawn along a horizontal circle.

## Turn the constraint into a distance measurement

Let $r$ denote the slant distance from the apex and $\alpha$ the angle between the cone's side and its central axis. A point at $r$ lies on a horizontal circle of radius $r\sin\alpha$. Small radial and azimuthal displacements on the surface meet at right angles. Their squared distance adds to $ds^2=dr^2+r^2\sin^2\alpha\,d\phi^2$. The change of variable $\psi=\phi\sin\alpha$ turns this into the polar metric of a plane. Cutting the cone along a generator and flattening it produces a sector of opening angle $2\pi\sin\alpha$.

Without a tangential force a particle follows a straight line in the unrolled sector. Its coordinates need not evolve linearly in the original azimuth. For two positions at equal slant distance $R$ separated by physical azimuth $\Delta\phi$ of magnitude at most $\pi$, the shortest branch uses a straight chord spanning $\Delta\psi=\sin\alpha\,\Delta\phi$. Bisecting the isosceles triangle gives

$$L=2R\sin(|\Delta\psi|/2),\qquad r_{\min}=R\cos(\Delta\psi/2).$$

The other copies of the endpoint correspond to winding around the cone on different branches. The chord formula uses a branch reached without passing through the apex. When the physical endpoints are diametrically opposite, clockwise and anticlockwise branches have equal shortest length. A laboratory string stretched across the interior would use the three-dimensional chord, which need not stay on the surface.

## Detect the departure from the horizontal ring

For a cone of half-angle 30 degrees, two marks 120 degrees apart at slant distance 5 cm appear only 60 degrees apart in the plane. Their shortest surface path is 5.00 cm long and passes through radius 4.33 cm. Staying on the horizontal ring would require $R\Delta\psi=5.24$ cm. The straight three-dimensional chord between the same marks is only 4.33 cm long but cuts through the cone. Comparing those three lengths distinguishes a free-space displacement from two distinct constrained motions.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
half_angle_deg = 30.
azimuth_deg = 120.
slant_cm = 5.
```

```python
assert 0 < half_angle_deg < 90 and 0 < azimuth_deg <= 180 and slant_cm > 0
s = np.sin(np.deg2rad(half_angle_deg))
delta = np.deg2rad(azimuth_deg)*s
edge = delta/2
closest = slant_cm*np.cos(edge)
length = 2*slant_cm*np.sin(edge)
assert length <= slant_cm*delta + 1e-12
print(f"Surface chord {length:.3f} cm, ring arc {slant_cm*delta:.3f} cm, closest radius {closest:.3f} cm")
```

```python
angles = np.linspace(-edge,edge,160)
fig, ax = axes("Unrolled radial-axis projection (cm)", "Unrolled transverse coordinate (cm)")
for end in [-edge,edge]:
    ax.plot([0,slant_cm*np.cos(end)],[0,slant_cm*np.sin(end)],color="#90999c",lw=1)
ax.plot(slant_cm*np.cos(angles),slant_cm*np.sin(angles),color=orange,label="Constant slant radius")
ax.plot([closest,closest],[-length/2,length/2],color=teal,lw=2.5,label="Shortest surface path")
ax.scatter([0],[0],color=blue,label="Apex");ax.set_aspect("equal",adjustable="box");ax.legend()
plt.show()
```

The endpoints share a radius while the middle of the geodesic lies closer to the apex. Both axes measure lengths intrinsic to the surface.

```python
sweep=np.linspace(0,180,181); phi=np.deg2rad(sweep)*s
fig, ax=axes("Physical separation (degrees of azimuth)", "Surface length (cm)")
ax.plot(sweep,2*slant_cm*np.sin(phi/2),color=teal,label="Geodesic length")
ax.plot(sweep,slant_cm*phi,"--",color=orange,label="Ring arc length")
ax.plot(sweep,slant_cm*np.cos(phi/2),color=blue,label="Closest slant radius")
ax.axvline(azimuth_deg,color="grey",ls=":");ax.legend()
plt.show()
```

Changing only the endpoint separation alters the sag towards the apex. At small separation the geodesic length approaches the circular arc.

## Exercise · infer the slope

Two marks at slant distance 10 cm and azimuth separation 120 degrees have a geodesic of length 10 cm. What is the cone half-angle?

The measured chord gives $\sin(\Delta\psi/2)=10/(2\cdot10)=1/2$. The shortest branch has $\Delta\psi=60^\circ$, hence $\sin\alpha=60/120=1/2$ and $\alpha=30^\circ$. Use the same angular units in the ratio.

## Exercise · compare with a weakly curved path

Find the relative excess of ring length over geodesic for a small unrolled separation $\delta$. Why does a very small patch look flat?

Expanding $2\sin(\delta/2)=\delta-\delta^3/24+\cdots$ gives $(R\delta-L)/(R\delta)=\delta^2/24+O(\delta^4)$. Both paths have the same tangent at leading order. The cone away from its apex is intrinsically developable; the small difference here measures the curvature of the chosen ring inside that developed plane.

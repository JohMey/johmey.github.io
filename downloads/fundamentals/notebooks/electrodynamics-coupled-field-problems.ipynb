# Weighing a small deposit with a piezoelectric resonator

Two nearby frequencies appear in the electrical response of one mechanical mode. Derive the series and parallel resonances, then use their different sensitivities to recognise added mass and stray capacitance.

## Convert mechanical motion into a terminal admittance

A piezoelectric plate converts terminal voltage into force and velocity into current. With a linear reciprocal coupling coefficient $n$, $F=nV$ and $I_m=nv$. The mechanical velocity impedance is $b+i\omega m+k/(i\omega)$. Dividing it by $n^2$ maps it to a series branch with $R_m=b/n^2$, $L_m=m/n^2$ and $C_m=n^2/k$. The electrodes also have a static capacitance $C_0$ in parallel. Thus

$$Y(\omega)=i\omega C_0+\left[R_m+i\omega L_m+\frac{1}{i\omega C_m}\right]^{-1}.$$

For negligible loss the series branch impedance vanishes at $\omega_s=1/\sqrt{L_mC_m}$. Just above this frequency its susceptance is inductive and can cancel the parallel capacitance. Setting $Y=0$ gives

$$
\omega_p=\omega_s\sqrt{1+C_m/C_0}.
$$

With finite damping a peak and a dip replace the singular limits. Their extrema need not coincide exactly with the lossless frequencies.

## Isolate the perturbation

A thin rigid deposit that changes only the effective mass multiplies $L_m$ by $1+\eta$. Both ideal frequencies scale as $(1+\eta)^{-1/2}$; their ratio is unchanged. Additional cable capacitance increases $C_0$, changing the ratio while leaving the mechanical series frequency fixed. At a load fraction of 0.01 the expected series shift is about minus 0.496 percent. A viscoelastic deposit can also dissipate energy and alter stiffness, so the two-frequency test includes a physical assumption about the coating.
For $f_s=100.000$ kHz, $C_m=0.25$ pF and $C_0=15$ pF, the unloaded parallel frequency is 100.830 kHz. A 1% increase in effective mass puts the series resonance at 99.504 kHz. If the same shift came from extra cable capacitance alone, the series frequency would remain at 100.000 kHz. At quality factor 600 the approximate unloaded resonance linewidth is $f_s/Q=167$ Hz. The predicted mass shift spans about three linewidths. The height of the peak by itself would confound damping and transduction.

Conversely, fit the two frequencies and infer the capacitance ratio as $(f_p/f_s)^2-1$. A small random error in $f_p/f_s$ has a large relative effect on this estimate when the resonances nearly coincide. Zoom the frequency grid and compare the uncertainty from bin spacing with the shift under investigation.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
series_hz = 100_000.
motional_pf = .25
parallel_pf = 15.
quality = 600.
mass_fraction = .01
```

```python
cm,c0=motional_pf*1e-12,parallel_pf*1e-12
assert min(cm,c0,series_hz,quality)>0 and mass_fraction>=0
lm=1/((2*np.pi*series_hz)**2*cm)
rm=2*np.pi*series_hz*lm/quality
def response(f,eta,c):
    w=2*np.pi*f
    return 1j*w*c+1/(rm+1j*w*lm*(1+eta)+1/(1j*w*cm))
fs=series_hz/np.sqrt(1+mass_fraction)
fp=fs*np.sqrt(1+cm/c0)
print(f"Loaded lossless frequencies {fs:.2f} Hz and {fp:.2f} Hz")
```

```python
f=np.linspace(.975*min(fs,series_hz),1.025*max(fp,series_hz*np.sqrt(1+cm/c0)),1200)
fig,ax=axes("Frequency (kHz)","Terminal admittance magnitude (µS)")
for eta,colour in [(0,teal),(mass_fraction,orange)]:
    ax.semilogy(f/1e3,abs(response(f,eta,c0))*1e6,label=f"Added mass {100*eta:g}%",color=colour)
ax.legend()
plt.show()
```

A terminal sweep contains the mechanical resonance and the electrical cancellation. Follow both features before inferring added mass.

```python
eta=np.linspace(0,.04,120)
fseries=series_hz/np.sqrt(1+eta)
fig,ax=axes("Series resonance (kHz)","Parallel resonance (kHz)")
for scale,colour in [(.7,orange),(1,teal),(1.5,blue)]:
    ax.plot(fseries/1e3,fseries*np.sqrt(1+cm/(scale*c0))/1e3,label=f"C₀ × {scale:g}",color=colour)
ax.legend()
plt.show()
```

Changing cable capacitance displaces the parallel resonance at fixed series frequency. Mass loading moves both frequencies along a diagonal path.

# Resolve two nearby wavelengths by watching fringes disappear

A Michelson interferometer shows very fine bright and dark fringes as one mirror moves. Two narrow spectral lines produce a much slower variation of the fringe contrast. Use the mirror travel between contrast minima to determine the separation of the lines.

## Combine two independent optical frequencies

Let $d$ be mirror displacement from equal paths. The optical path difference of the returning beams is $2d$ and the phase at vacuum wavenumber $\sigma=1/\lambda$ is $4\pi\sigma d$. Assume equal arm intensities for each individual line and an ideal detector that averages quickly oscillating optical fields and the beat between the distinct source frequencies. The measured intensities from the two spectral components add.

If the second line carries a fraction $r$ of the first line's intensity, the modulated part of the signal divided by the total mean becomes

$$
s=[\cos(4\pi\sigma_1d)+r\cos(4\pi\sigma_2d)]/(1+r).
$$

Writing the two cosines as real parts separates a rapidly rotating phase from the complex envelope. Taking the magnitude produces the normalised visibility,

$$V(d)=\frac{\sqrt{1+r^2+2r\cos(4\pi\Delta\sigma d)}}{1+r},\qquad \Delta\sigma=|\sigma_1-\sigma_2|. $$

This is the visibility obtained by scanning enough fine fringes to estimate local maxima and minima whilst the envelope changes negligibly. For equally strong lines it simplifies to $|\cos(2\pi\Delta\sigma d)|$. The first loss of contrast occurs at $d=1/(4\Delta\sigma)$ and successive minima are spaced by $1/(2\Delta\sigma)$. The factor of two relative to optical path difference comes directly from reflection at the moving mirror.

## Compare mirror travel with a single optical fringe

Choose vacuum wavelengths 589.0 and 589.6 nm. Their wavenumber separation is $\Delta\sigma=0.6/(589.0\times589.6)$ nm⁻¹, about $1.73\times10^3$ m⁻¹. The first visibility minimum lies at mirror displacement 0.145 mm and contrast revives at 0.289 mm. An individual mirror fringe spans only about 0.295 µm. Thus hundreds of carrier fringes fit inside a single envelope lobe. A coarse plot of raw intensity over a millimetre can alias those fast oscillations.

Suppose the second line carries half the intensity of the first. At the same minimum the two modulation vectors point in opposite directions. The surviving magnitude relative to their summed intensities is $|1-0.5|/(1+0.5)=1/3$. An incomplete minimum can therefore arise even when both components are individually sharp. A finite width of either line or imperfect overlap of the arms in a real interferometer can also reduce the observed contrast. Independent measurements near equal path help identify such instrumental loss.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
wavelength1_nm = 589.0
wavelength2_nm = 589.6
relative_line_intensity = 1.0
```

```python
assert wavelength1_nm>0 and wavelength2_nm>0 and wavelength1_nm!=wavelength2_nm
assert relative_line_intensity>0
sigma1,sigma2 = 1/(wavelength1_nm*1e-9), 1/(wavelength2_nm*1e-9)
separation = abs(sigma1-sigma2)
first_min = 1/(4*separation)
r = relative_line_intensity
def detector(d):
    return 1+(np.cos(4*np.pi*sigma1*d)+r*np.cos(4*np.pi*sigma2*d))/(1+r)

def visibility(d,r):
    return np.abs(1+r*np.exp(4j*np.pi*separation*d))/(1+r)

assert np.isclose(visibility(first_min,r),abs(1-r)/(1+r))
assert np.isclose(visibility(2*first_min,r),1.)
print(f"First minimum at {first_min*1e3:.6f} mm; minimum visibility {visibility(first_min,r):.4f}")
```

```python
fine = np.linspace(-2,2,901)/(sigma1+sigma2)
fig,ax = axes("Mirror displacement about the window centre (µm)", "Intensity / mean intensity")
for centre,label,color in [(0,"Equal paths",teal),(first_min,"First contrast minimum",orange)]:
    ax.plot(fine*1e6,detector(centre+fine),color=color,label=label)
ax.set_ylim(-.05,2.05);ax.legend()
plt.show()
```

Each curve shows a short scan around the position indicated in its legend. Only the window near equal paths gives full contrast when the lines have equal intensity.

```python
mirror = np.linspace(0,6*first_min,700)
fig,ax = axes("Mirror displacement from equal paths (mm)", "Local visibility")
ax.plot(mirror*1e3,visibility(mirror,r),color=teal,label=f"Selected ratio {r:g}")
comparison = .5 if not np.isclose(r,.5) else 1.
ax.plot(mirror*1e3,visibility(mirror,comparison),"--",color=orange,label=f"Comparison ratio {comparison:g}")
ax.axvline(first_min*1e3,color="grey",ls=":",label="First minimum")
ax.set_ylim(-.03,1.08);ax.legend()
plt.show()
```

The separation between minima determines wavenumber splitting even when unequal line intensities prevent complete cancellation.

## Exercise · infer a line separation from successive minima

A mirror moves 0.400 mm between two consecutive contrast minima. Estimate the wavelength separation for lines centred near 600 nm.

The wavenumber difference is $\Delta\sigma=1/(2\times0.400\,\mathrm{mm})=1250$ m⁻¹. The narrow-doublet approximation $|\Delta\lambda|\simeq\lambda^2\Delta\sigma$ gives $0.450$ nm. Using the optical path difference as though it were mirror displacement would introduce a factor of two.

## Exercise · infer the weaker line fraction

An ideal instrument observes maximum visibility one and minimum visibility 0.20. How strong is the weaker line relative to the stronger one? Can visibility label which line has the longer wavelength?

Calling the weaker-to-stronger ratio $r\le1$ gives $(1-r)/(1+r)=0.20$, hence $r=2/3$. The visibility magnitude is symmetric when the two lines are exchanged. Assigning the stronger intensity to a particular wavelength requires additional spectral information.

# When does a smooth circular guide lose contact?

A bead moves inside a vertical circular guide after receiving a known speed at the bottom. Predict whether it turns around, separates from the guide or completes the circle. A normal-force measurement can distinguish these outcomes before a simple top-height energy test can.

## Check the constraint force after solving for speed

Use $\theta=0$ at the bottom, increasing along the rising side. Let $R$ be the radius of the bead centre's path, $v_0$ its bottom speed and $u=v_0^2/(gR)$. The smooth guide does no work. The rise in gravitational energy $mgR(1-\cos\theta)$ therefore gives $v^2/(gR)=u-2+2\cos\theta$. This relation applies only while the bead follows the circle and the computed speed is real.

With inward radial direction positive, force balance is $N-mg\cos\theta=mv^2/R$. Substitution gives

$$\frac{N}{mg}=u-2+3\cos\theta.$$

A guide underneath the bead can push but cannot pull. For $u<2$ the speed first reaches zero while the contact force is still positive. For $2<u<5$ the force reaches zero with positive speed at $\cos\theta=(2-u)/3$. Above that point circular motion would require a negative normal force. For $u>5$ the bead can remain in contact throughout the circle. At the boundaries $u=2$ gives a zero-speed, zero-normal grazing turn at the side and $u=5$ gives marginal contact at the top. A captive slotted carriage or a bead threaded onto a wire has a different constraint and can transmit force of either sign.

## Distinguish an energetically accessible top from a feasible trajectory

Let $R=0.25$ m and choose $v_0=\sqrt{3gR}=2.71$ m/s. The guide loses contact when $\cos\theta=-1/3$, or at 109.5 degrees from the bottom. The remaining speed there is $\sqrt{gR/3}=0.904$ m/s. Raising the initial energy so that $u=4$ pays the gravitational energy cost to reach the top, but a continuously guided motion still fails the nonnegative force condition earlier. The mechanical-energy and force-balance tests answer different questions.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
radius_m=.25
speed_parameter=3.0    # v0² / gR
gravity=9.81
```

```python
assert radius_m>0 and gravity>0 and 0<speed_parameter<=8
def endpoint(u):
    if u<=2: return np.arccos(1-u/2),"turn"
    if u<5: return np.arccos((2-u)/3),"detachment"
    return np.pi,"top reached"
end,outcome=endpoint(speed_parameter)
bottom=np.sqrt(speed_parameter*gravity*radius_m)
v_end=np.sqrt(gravity*radius_m*max(0,speed_parameter-2+2*np.cos(end)))
print(f"Bottom speed {bottom:.3f} m/s; {outcome} at {np.rad2deg(end):.3f}°, speed {v_end:.3f} m/s")
```

```python
angle=np.linspace(0,end,170); cs=np.cos(angle)
fig,ax=axes("Angle from the bottom (degrees)","Dimensionless quantity")
ax.plot(np.rad2deg(angle),speed_parameter-2+3*cs,color=teal,label="Normal force / mg")
ax.plot(np.rad2deg(angle),speed_parameter-2+2*cs,color=orange,label="Squared speed / gR")
ax.axhline(0,lw=.7,color="grey");ax.set_xlim(0,180);ax.legend()
plt.show()
```

The first quantity to reach zero determines which constraint fails. The part after contact loss requires a different trajectory.

```python
us=np.linspace(.01,7,240); ends=[np.rad2deg(endpoint(val)[0]) for val in us]
fig,ax=axes("Bottom speed parameter v₀² / gR","Last guided angle on ascent (degrees)")
ax.plot(us,ends,color=teal);ax.axvline(2,color=blue,ls="--",label="Grazing turn")
ax.axvline(5,color=orange,ls=":",label="Marginal top contact");ax.plot(speed_parameter,np.rad2deg(end),"o",color=orange);ax.legend()
plt.show()
```

The horizontal branch records completion of the ascent, including marginal contact at its threshold. It does not mean that the bead stops at the top.

## Exercise · find the lower turning point

For $u=1$, determine the turning angle and force there.

Zero speed requires $\cos\theta=1-u/2=1/2$, hence $\theta=60^\circ$. The contact force then balances the outward radial component of gravity, $N=mg\cos\theta=mg/2$. It remains compressive through the turn.

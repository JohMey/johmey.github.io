# Extract work as two finite bodies equilibrate

Two bodies in an insulated enclosure begin at different temperatures. A cyclic reversible device could extract work before their temperatures equalise. Compare this with simply putting the bodies in thermal contact. Change the heat capacities and initial temperatures, then rerun.

```python
import numpy as np
import matplotlib.pyplot as plt

Th,Tc=360.,280.         # initial temperatures, K
Ch,Cc=1.2,1.2           # total heat capacities, kJ/K
assert Th>Tc>0 and Ch>0 and Cc>0
```

## Close separate entropy and energy balances
A reversible final state has the initial total entropy. Direct contact preserves the initial total energy. Use these two balances separately to obtain the two different final temperatures.

```python
Trev=np.exp((Ch*np.log(Th)+Cc*np.log(Tc))/(Ch+Cc))
Tmix=(Ch*Th+Cc*Tc)/(Ch+Cc)
W=(Ch+Cc)*(Tmix-Trev)
deltaS=Ch*np.log(Tmix/Th)+Cc*np.log(Tmix/Tc)
assert W>0 and deltaS>0
assert abs(Ch*np.log(Trev/Th)+Cc*np.log(Trev/Tc))<1e-11
print(f"Direct contact {Tmix:.3f} K; reversible end state {Trev:.3f} K; maximum work {W:.3f} kJ")
print(f"Entropy created in direct contact {deltaS:.5f} kJ/K.")
```

## Follow the reversible path
Transfer equal entropy increments out of the hot body and into the cold one. Work is the difference between the released and absorbed energies. At equal temperatures the marginal work vanishes.

```python
ds=np.linspace(0,Ch*np.log(Th/Trev),250)
hot=Th*np.exp(-ds/Ch);cold=Tc*np.exp(ds/Cc)
work=Ch*(Th-hot)-Cc*(cold-Tc)
assert np.isclose(work[-1],W) and np.isclose(hot[-1],cold[-1])
fig,ax=plt.subplots(1,2,figsize=(9,3.5),layout="constrained")
ax[0].plot(ds,hot,label="Hot body");ax[0].plot(ds,cold,label="Cold body")
ax[0].set(xlabel="Entropy transferred (kJ/K)",ylabel="Temperature (K)",title="Equal entropy exchanged")
ax[1].plot(ds,work,label="Work delivered")
ax[1].axhline(W,ls=":",color="0.45",label="Maximum")
ax[1].set(xlabel="Entropy transferred (kJ/K)",ylabel="Accumulated work (kJ)",title="End of the work-producing path")
for p in ax:p.grid(alpha=.22);p.legend(fontsize=8)
plt.show()
```

# Where does a circular opening concentrate stress?

Make a hole in a flat sheet pulled along one direction. Derive the conditions on the empty edge, compute the circumferential stress and check how quickly the disturbance decays as a strain gauge moves into the sheet.

## Keep the free edge and the distant traction in one solution

Take an isotropic linear-elastic sheet, long and wide relative to a circular through-hole of radius $a$. At a large distance apply uniaxial tension $S$ along $x$. The empty edge can carry neither normal nor tangential traction, so $\sigma_{rr}=\tau_{r\theta}=0$ at $r=a$. The circumferential stress $\sigma_{\theta\theta}$ is parallel to this edge and need not vanish. Writing uniaxial tension in polar coordinates introduces only a constant angular sector and a sector in $\cos(2\theta)$ and $\sin(2\theta)$.

In plane elasticity an Airy stress function satisfies equilibrium automatically and must be biharmonic for a homogeneous isotropic solid without body forces. The circular-hole solution can be built by adding decaying terms in those same two angular sectors to the distant uniform stress. One convenient potential is

$$\Phi=\frac{S}{4}\left[r^2-2a^2\log(r/a)-(r^2-2a^2+a^4/r^2)\cos(2\theta)\right].$$

Differentiate twice with respect to radius to obtain the hoop stress. Writing $u=(a/r)^2$ gives

$$
\sigma_{\theta\theta}=\frac S2[1+u-(1+3u^2)\cos(2\theta)].
$$

The radial and shear stress expressions in the calculation cell follow from $\sigma_{rr}=\Phi_r/r+\Phi_{\theta\theta}/r^2$ and $\tau_{r\theta}=-\partial_r(\Phi_\theta/r)$. Check both edge tractions explicitly; imposing zero hoop stress instead would erase the stress concentration we aim to measure.

## Locate a gauge and interpret its reading

On the edge $u=1$, hence $\sigma_{\theta\theta}=S[1-2\cos(2\theta)]$. With $S=40$ MPa, gauges at the top and bottom of the opening align with tensile hoop stress of 120 MPa. At the point facing along the remote pulling direction the hoop stress is compressive at −40 MPa. The nominal far-field stress refers to the gross, undisturbed section. A factor based on a finite specimen’s reduced net section would use a different reference.

Along the transverse radius, moving from $r=a$ to $r=2a$ reduces the ratio to $1+u/2+3u^2/2=1.21875$. A gauge whose active length samples a substantial fraction of the hole radius therefore averages a gradient. Failure to read a threefold concentration at a distant or finite-area gauge does not by itself invalidate the elastic solution.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
radius_mm = 8.0
remote_mpa = 40.0
```

```python
assert radius_mm>0 and remote_mpa>0
def stresses(r_over_a,theta):
    u = np.asarray(r_over_a,dtype=float)**-2
    c,s = np.cos(2*theta),np.sin(2*theta)
    rr = .5*remote_mpa*(1-u+(1-4*u+3*u*u)*c)
    tt = .5*remote_mpa*(1+u-(1+3*u*u)*c)
    rt = -.5*remote_mpa*(1+2*u-3*u*u)*s
    return rr,tt,rt
ang=np.linspace(0,2*np.pi,181)
rr,tt,rt=stresses(1.,ang)
assert np.allclose(rr,0) and np.allclose(rt,0)
print(f"Peak hoop stress {tt.max():.2f} MPa; transverse reading at two radii {stresses(2.,np.pi/2)[1]:.2f} MPa")
```

```python
coord=np.linspace(-3,3,301)
X,Y=np.meshgrid(coord,coord); rad=np.hypot(X,Y)
hoop=np.ma.masked_where(rad<1,stresses(np.maximum(rad,1),np.arctan2(Y,X))[1])
fig,ax=axes("x (mm)","y (mm)")
im=ax.contourf(radius_mm*X,radius_mm*Y,hoop,levels=24,cmap="coolwarm")
fig.colorbar(im,ax=ax,label="Hoop stress (MPa)")
ax.set_aspect("equal")
plt.show()
```

The white interior represents removed material. The colour field is one stress component expressed relative to the local circumferential direction.

```python
fig,ax=axes("Distance from hole edge (mm)","Circumferential stress (MPa)")
r=np.linspace(1,5,301)
for th,label,col in [(0,"Along the loading direction",orange),(np.pi/2,"Across the loading direction",teal)]:
    ax.plot(radius_mm*(r-1),stresses(r,th)[1],label=label,color=col)
ax.axhline(0,color="0.5",lw=.8);ax.legend()
plt.show()
```

The horizontal coordinate measures actual distance outward from the wall of the hole. Moving the gauge outward changes the inferred peak.

## Exercise · change the radius

Double the hole radius at the same remote stress in an effectively infinite sheet. What changes in the peak and the stress gradient in MPa per millimetre?

The peak remains $3S$ because the stress field depends only on $r/a$ and angle. The physical length scale doubles, hence an otherwise corresponding spatial gradient halves. Finite edge distances, plastic yielding and rounded specimen boundaries can introduce new scales and change these conclusions.

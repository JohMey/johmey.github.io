# Can an atom leave its initial state and still pass the final test?

A driven two-level atom undergoes repeated ideal measurements. Compare two laboratory records, one retaining only runs that passed every test and another retaining every run. The distinction is visible even when every measurement is made with the same apparatus.

## Count histories or evolve a population

Write the resonant Hamiltonian in the rotating frame as $H=\hbar\Omega\sigma_x/2$. A pulse of duration $T/N$ rotates the initial ground state into $\cos(\alpha/2)|0\rangle-i\sin(\alpha/2)|1\rangle$ with $\alpha=\Omega T/N$. The probability of passing one ground-state test is $c=\cos^2(\alpha/2)$. Each successful projection prepares the same ground state for the next interval. The probability that a complete record contains only successes is consequently $P_{\rm all}=c^N$.

For an unread measurement, keep the atoms detected in either state. The projection removes the off-diagonal density-matrix elements, leaving a classical mixture just before the next pulse. If $p_j$ denotes the ground fraction after measurement $j$, the next pulse maps it to $p_{j+1}=c p_j+(1-c)(1-p_j)$. The excess population $z_j=2p_j-1$ then obeys $z_{j+1}=(2c-1)z_j$. Starting in the ground state gives

$$P_{\rm final}=\frac{1+\cos^N(\Omega T/N)}2. $$

Leaving the ground state in one interval and returning in a later one contributes to the final fraction. It removes that run from the all-success record. The two probabilities therefore refer to different experimental questions, even though they use exactly the same sequence of pulses and measurements.

## Follow a half-turn in two steps

Choose $\Omega T=\pi$ and make two evenly spaced measurements, including the one at the end. Each pulse is a quarter-turn of the Bloch vector and gives either result with equal probability following either previous outcome. The four two-result strings have probability one quarter each. Only 00 passes every test, whereas 00 and 10 both finish in the ground state. The fractions are therefore 0.25 and 0.50. With a single final measurement the undisturbed half-turn ends entirely in the excited state.

At fixed total duration and very large $N$, Taylor expansion gives $\log P_{\rm all}\simeq-(\Omega T)^2/(4N)$. The survival probability tends to one because the transition probability per interval is quadratic in its short duration. This prediction assumes ideal instantaneous projective measurements and coherent rotations between them. A measurement with a fixed finite duration cannot be packed arbitrarily densely into a fixed time.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
pulse_area_in_pi = 1.0
measurements = 8
```

```python
assert measurements >= 1 and int(measurements) == measurements
measurements = int(measurements)
area = np.pi * pulse_area_in_pi
def records(area, count):
    c = np.cos(area / (2*count))**2
    return c**count, (1 + np.cos(area/count)**count)/2

all_pass, final_ground = records(area, measurements)
angle = area / (2*measurements)
rotation = np.array([[np.cos(angle), -1j*np.sin(angle)], [-1j*np.sin(angle), np.cos(angle)]])
rho = np.diag([1., 0.])
for j in range(measurements):
    rho = np.diag(np.diag(rotation @ rho @ rotation.conj().T))
assert np.isclose(rho[0,0], final_ground)
print(f"Passed every test: {all_pass:.6f}. Ground at the final test: {final_ground:.6f}.")
```

```python
step = np.arange(measurements+1)
time_fraction = np.linspace(0,1,401)
fig, ax = axes("Time / total drive time", "Probability")
ax.plot(time_fraction,np.cos(area*time_fraction/2)**2,color="grey",label="Uninterrupted drive")
ax.plot(step/measurements,np.cos(angle)**(2*step),"o-",color=teal,label="Passed every test")
ax.plot(step/measurements,(1+np.cos(2*angle)**step)/2,"s--",color=orange,label="Ground at this test")
ax.set_ylim(-.03,1.06); ax.legend()
plt.show()
```

Markers represent measurements. The continuous reference curve is the ground population of an atom subjected to the same drive with every intermediate test omitted.

```python
counts = np.arange(1,max(51,min(500,measurements+1)))
series = records(area,counts)
fig,ax = axes("Number of measurements in the fixed total time", "Probability")
ax.plot(counts,series[0],".-",color=teal,label="Passed every test")
ax.plot(counts,series[1],"--",color=orange,label="Ground at final test")
ax.set_ylim(-.03,1.06);ax.legend()
plt.show()
```

Keep the pulse area and total time fixed whilst shortening each free interval. An increasingly dense train of ideal observations preserves the initial state.

## Exercise · recover a mixed population

If the atom initially occupies the ground state with probability 0.7 and has no coherences, find the final ground probability for four measurements and a total pulse area of $\pi$.

The initial contrast is 0.4. Each interval multiplies it by $\cos(\pi/4)=1/\sqrt2$. Four steps give contrast 0.1, so the final ground fraction is $(1+0.1)/2=0.55$. The all-pass probability would instead contain a first measurement conditioned on the initially mixed preparation.

Estimate the number of ideal tests needed for 99 percent all-pass survival during a coherent half-turn. Check the estimate with the editable input.

# Which damping law produced the free-decay trace?

A carriage slides on a guide and a spring pulls it back to its resting position. Two dampers can be adjusted to give the same first peak reduction. Predict the following turning points and decide which measurement separates viscous loss from dry sliding friction.

## Use the turning points to identify the energy loss mechanism

For a viscous damper the free equation is $m\ddot x+c\dot x+kx=0$. In the underdamped regime let $\omega_n=\sqrt{k/m}$, $\zeta=c/(2\sqrt{km})$ and $\omega_d=\omega_n\sqrt{1-\zeta^2}$. Released from $x(0)=A_0$ with zero velocity, the displacement is $A_0e^{-\zeta\omega_nt}[\cos\omega_dt+(\zeta\omega_n/\omega_d)\sin\omega_dt]$. Turning points occur every $\pi/\omega_d$. Their absolute sizes have a constant ratio $e^{-\pi\zeta/\sqrt{1-\zeta^2}}$.

Now let a dry guide exert a constant opposing force of magnitude $F$. During a sliding half-cycle its sign remains fixed. Starting from a positive peak with $kx>F$, the spring accelerates the mass to the left while friction acts to the right. Subtract the shifted equilibrium $F/k$ from the coordinate. The remainder obeys an undamped oscillator equation, so the next turning point after $\pi/\omega_n$ is $2F/k-x$. For a release from the other side, replace $F/k$ by its signed counterpart. The magnitudes of successive peaks fall by $2F/k$ while the mass still crosses the origin. At a turning point where $|kx|\le F$, static friction can support the spring and the body sticks there. Here the static threshold is idealised as equal to the sliding force.

## Match the first peak and predict the next three

Let $m=1$ kg, $k=1000$ N/m, $F=0.6$ N and $A_0=12$ mm. Dry friction subtracts 1.2 mm of magnitude per sliding half-cycle. The first return is at −10.8 mm. A viscous model can reproduce its magnitude if its half-period peak ratio is $0.9$. Solving $\delta=\pi\zeta/\sqrt{1-\zeta^2}$ for $\delta=-\ln(0.9)$ gives $\zeta=0.0335$. Its fourth half-cycle peak is $12(0.9)^4=7.87$ mm, whereas the dry-friction prediction is 7.20 mm. A single peak pair cannot distinguish the models. Several peaks well above the instrument resolution can.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
mass = 1.0
stiffness = 1000.0
friction = 0.6
initial_mm = 12.0
```

```python
A=initial_mm/1000.; a=friction/stiffness; wn=np.sqrt(stiffness/mass)
assert mass>0 and stiffness>0 and 0<2*a<A
first_ratio=1-2*a/A
delta=-np.log(first_ratio)
zeta=delta/np.hypot(np.pi,delta)
wd=wn*np.sqrt(1-zeta*zeta)
states=[A]
while abs(states[-1])>a*(1+1e-10):
    x0=states[-1]; states.append(2*np.sign(x0)*a-x0)
    assert len(states)<10001, "Select a resolvable sliding friction."
states=np.array(states)
assert np.isclose(abs(states[1]),A*first_ratio) and abs(states[-1])<=a*(1+1e-10)
print(f"Matched viscous ratio {zeta:.4f}; sliding ends after {len(states)-1} half cycles at {1e3*states[-1]:.3f} mm.")
```

```python
t=np.linspace(0,max(.3,(len(states)+2)*np.pi/wn),1001)
j=np.minimum((t/(np.pi/wn)).astype(int),len(states)-1)
last=j==len(states)-1
eq=np.sign(states[j])*a
dry=np.where(last,states[-1],eq+(states[j]-eq)*np.cos(wn*(t-j*np.pi/wn)))
vis=A*np.exp(-zeta*wn*t)*(np.cos(wd*t)+zeta*wn/wd*np.sin(wd*t))
fig,ax=axes("Time (s)","Displacement (mm)")
ax.plot(t,vis*1000,label="Viscous",color=teal);ax.plot(t,dry*1000,"--",label="Dry sliding and sticking",color=orange)
ax.legend();plt.show()
```

The sliding model is evaluated on exact half-cycle segments. A plateau is physical when the static force threshold can hold the carriage.

```python
fig,ax=axes("Half-cycle turning-point index","Absolute turning point (mm)")
j=np.arange(len(states))
ax.plot(j,1000*np.abs(states),"o-",label="Dry friction",color=orange)
ax.plot(j,1000*A*first_ratio**j,"s--",label="Viscous",color=teal)
ax.legend();plt.show()
```

Plotting amplitude against peak count makes an almost missed difference in the first displacement trace easier to measure.

## Exercise · check the last interval

A positive turning point is 0.9 mm and $F/k=0.6$ mm. Will the mass reach a negative turning point?

Sliding restarts because $0.9>0.6$. The next position is $2(0.6)-0.9=0.3$ mm, still positive. Since $0.3<0.6$, static friction can hold it there. Extrapolating a positive linear amplitude envelope through zero would lose the actual final position.

# Detecting the Gouy phase at a beam waist
Compare focused Gaussian propagation with a plane reference carrying the same axial phase. Predict how a larger waist changes the location and total amount of the residual phase shift.

```python
import numpy as np
import matplotlib.pyplot as plt

waist_um=40.
wavelength_um=.633
extent_in_rayleigh=4.
reference_offset_rad=.4
assert waist_um>3*wavelength_um>0 and extent_in_rayleigh>0
```

## Recover phase from two interference quadratures
Use two calibrated equal-amplitude reference measurements to reconstruct the phase independent of amplitude. The reference sign is defined so the on-axis field phase is kz minus the Gouy angle.

```python
zr=np.pi*waist_um**2/wavelength_um
s=np.linspace(-extent_in_rayleigh,extent_in_rayleigh,401);z_mm=zr*s/1000
envelope=1/(1+1j*s)
phase=np.unwrap(np.angle(envelope))
c=np.real(envelope*np.exp(1j*reference_offset_rad))
q=np.imag(envelope*np.exp(1j*reference_offset_rad))
recovered=np.unwrap(np.arctan2(q,c))-reference_offset_rad
assert np.allclose(np.exp(1j*recovered),np.exp(1j*phase))
print("Rayleigh range (mm):",zr/1000,"; residual phase from −zR to +zR (rad):",-np.pi/2)
```

## Separate axial beam width from axial phase
Reflecting z about the waist preserves beam size and reverses the phase shift. An intensity-only scan cannot distinguish the two signs. The two readout channels on the right retain sign information after calibration.

```python
fig,ax=plt.subplots(1,2,figsize=(9,3.9),layout="constrained")
ax[0].plot(z_mm,phase,color="#175868",label="Gaussian residual")
ax[0].axhline(0,ls=":",color="grey",label="Plane-wave residual");ax[0].legend(fontsize=8)
ax[0].set(xlabel="Axial coordinate (mm)",ylabel="Phase after removing kz (rad)",title="On-axis Gouy phase")
ax[1].plot(z_mm,c,color="#375b9a",label="In-phase channel")
ax[1].plot(z_mm,q,color="#a23d21",label="Quadrature channel")
ax[1].set(xlabel="Axial coordinate (mm)",ylabel="Normalised field quadrature",title="Two-phase interferometer");ax[1].legend(fontsize=8)
plt.show()
```

# Put a thermometer on a bounded spin spectrum

A collection of identical two-level degrees of freedom has both a lowest and a highest energy. Count the accessible states at each energy and determine from adjacent entropy differences which way energy flows when two systems are brought into weak thermal contact.

## Differentiate an entropy measured in counts

Let $N$ distinguishable, weakly coupled sites each carry energy zero or $\varepsilon$. A state with $n$ excitations has $E_n=n\varepsilon$. The multiplicity of the narrow, fixed-energy shell equals the number of ways to place the excitations, $W_n={N\choose n}$. Use the Boltzmann shell entropy $S_n=k_B\ln W_n$. The temperature inferred from the population of two adjacent shells samples the derivative at an energy midway between them. Dividing their multiplicities before taking a logarithm gives an exact finite-system expression,

$$\beta_{n+1/2}\varepsilon=\frac{S_{n+1}-S_n}{k_B}=\ln\frac{N-n}{n+1},\qquad \beta=\frac{1}{k_BT}.$$

Adding an excitation near the bottom initially opens more arrangements. Beyond the middle, it closes them. The decreasing entropy on this upper branch assigns a negative inverse temperature to the population. In the limit of large $N$ at fixed excitation fraction $p$, the slope becomes $\ln[(1-p)/p]$. The same relationship follows from $p/(1-p)=\exp(-\beta\varepsilon)$ for a canonical two-level system.

The upper energy bound allows a distribution weighted towards high energy to normalise. In experiment the degrees of freedom considered must also equilibrate amongst themselves on a timescale shorter than their relaxation into ordinary, effectively unbounded environmental modes. This calculation describes the spin subsystem during that interval.

## Predict the direction of an exchange

Take $N=20$ and examine the transition from 15 to 16 excitations. The multiplicity ratio is $5/16$, the dimensionless inverse temperature is −1.1632, and $k_BT/\varepsilon$ is approximately −0.860. The reflected transition from 4 to 5 has ratio $16/5$ and the opposite slope. The same number of possible arrangements at energies reflected about the midpoint gives opposite entropy gradients.

Suppose system A initially has 16 excitations and the same-sized system B has 4. Moving one quantum from A to B multiplies the product of their multiplicities by $(16/5)^2=10.24$. A weak contact that samples the available microstates favours transfer out of the inverted specimen. More generally, when A gives up a small positive energy $q$, the entropy change is $\Delta S/k_B\simeq(\beta_B-\beta_A)q$. An inverted subsystem therefore gives up heat to an ordinary positive-temperature subsystem when the coupling permits energy exchange.

## Set the experimental inputs

Edit these values, then run the calculation and both figures again.

```python
import numpy as np
import matplotlib.pyplot as plt

teal, orange, blue = "#19596a", "#aa462a", "#3868a0"
def axes(xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)
    ax.set(xlabel=xlabel, ylabel=ylabel)
    ax.grid(alpha=0.2)
    return fig, ax
```

```python
sites = 20
lower_excitation = 15
```

```python
from math import lgamma
assert isinstance(sites,int) and sites >= 2
assert isinstance(lower_excitation,int) and 0 <= lower_excitation < sites
n = np.arange(sites+1)
log_count = np.array([lgamma(sites+1)-lgamma(int(j)+1)-lgamma(sites-int(j)+1) for j in n])
midpoint = (n[:-1]+.5)/sites
slope = np.log((sites-n[:-1])/(n[:-1]+1))
assert np.allclose(np.diff(log_count), slope)
assert np.allclose(slope, -slope[::-1])
print(f"Midpoint inverse temperature βε: {slope[lower_excitation]:.5f}")
```

```python
fig,ax = axes("Fraction of sites excited", "Shell entropy per site (in units of kB)")
ax.plot(n/sites,log_count/sites,".-",color=teal)
i=lower_excitation
ax.plot(n[i:i+2]/sites,log_count[i:i+2]/sites,"o-",color=orange,lw=3,label="Selected adjacent shells")
ax.legend()
plt.show()
```

The segment joining the two highlighted shells measures the entropy difference used in the finite-system inverse temperature.

```python
p = np.linspace(.015,.985,351)
fig,ax = axes("Excitation fraction at the energy midpoint", "Inverse temperature βε")
ax.plot(p,np.log((1-p)/p),"--",color=blue,label="Large-system curve")
ax.plot(midpoint,slope,".",color=teal,label="Finite shell differences")
ax.plot(midpoint[i],slope[i],"o",color=orange)
ax.axhline(0,color="grey",lw=.8);ax.legend()
plt.show()
```

The discrete slope belongs between neighbouring energies. Compare finite sizes at the same fractional position to see where the large-system approximation converges first.

## Exercise · the temperature halfway up an odd spectrum

For eleven sites find the adjacent shells with exactly zero inverse temperature. What site occupation does this represent in the canonical description?

Setting $11-n=n+1$ gives $n=5$. Five and six excitations have the same multiplicity, giving zero entropy slope at $E=5.5\varepsilon$. In the canonical ensemble $\beta=0$ gives equal probabilities for both site energies. The average excitation number is 5.5 although individual measurements return integers.

## Exercise · infer a population inversion

A spin measurement yields three excited sites for every site in the lower state, in a sample large enough to neglect finite-size corrections. Calculate the inverse temperature and the mean energy of 100 equivalent sites.

Here $p=3/4$. The Boltzmann ratio gives $\beta\varepsilon=-\ln3\simeq-1.099$, and the mean energy is

$$
100p\varepsilon=75\varepsilon.
$$

At the positive inverse temperature $+\ln3/\varepsilon$ the mean falls to $25\varepsilon$. The two occupations have the same binary mixing entropy and opposite response to an infinitesimal energy exchange.
